```python
#SIMPLE CODE FOR DICTIONARIES
student = {'name': 'ABDULLAH', 'age' : 22, 'subjects':['MATHS', 'COPMUTER', 'HISTORY']}
print(student)
```

```python
#GET CODE FOR DICTIONARIES
student = {'name': 'ABDULLAH', 'age' : 22, 'subjects':['MATHS', 'COPMUTER', 'HISTORY']}
print(student.get('phone', 'Not Found'))
```

```python
#ADD CODE FOR DICTIONARIES
student = {'name': 'ABDULLAH', 'age' : 22, 'subjects':['MATHS', 'COPMUTER', 'HISTORY']}
student['phone'] = '66666-5555-78'
print(student.get('phone', 'Not Found'))
```

```python
#UPDATINF THE OLD ONE WITH NEW ONE
student = {'name': 'ABDULLAH', 'age' : 22, 'subjects':['MATHS', 'COPMUTER', 'HISTORY']}
student['phone'] = '66666-5555-78'
student['name'] = 'BAHADUR'
print(student)
```

```python
#DELELTING
student = {'name': 'ABDULLAH', 'age' : 22, 'subjects':['MATHS', 'COPMUTER', 'HISTORY']}

del student['age']
print(student)
```

```python
#POP
student = {'name': 'ABDULLAH', 'age' : 22, 'subjects':['MATHS', 'COPMUTER', 'HISTORY']}
age = student.pop('age')
print(student)
print(age)
```

```python
#length in variable
student = {'name': 'ABDULLAH', 'age' : 22, 'subjects':['MATHS', 'COPMUTER', 'HISTORY']}
print(len(student))
```

```python
#VALUES
student = {'name': 'ABDULLAH', 'age' : 22, 'subjects':['MATHS', 'COPMUTER', 'HISTORY']}
print(student.values())
```

```python
#ITEMS
student = {'name': 'ABDULLAH', 'age' : 22, 'subjects':['MATHS', 'COPMUTER', 'HISTORY']}
print(student.items())
```

```python
#key
student = {'name': 'ABDULLAH', 'age' : 22, 'subjects':['MATHS', 'COPMUTER', 'HISTORY']}
for key in student.items():
  print(key)
```

```python
#key & value
student = {'name': 'ABDULLAH', 'age' : 22, 'subjects':['MATHS', 'COPMUTER', 'HISTORY']}
for key, value in student.items():
  print(key, value)
```
