```python
#THIS LOPPS CAN USED FOR ALL(CHR,INT,FLOT) (FOR LOOP)
nums = [1,2,3,4,5]
for num in nums:
   print(num)
```

```python
#WHEN THE GOAL IS FOUNDED LOOP STOPPED & BREAK FUCTION (FOR LOOP)
nums = [1,2,3,4,5]
for num in nums:
   if num == 4:
    print('FOUNDED!')
    break
   print(num)
```

```python
#LOOP WITH IN LOOP (FOR LOOP)
nums = [1,2,3]
for num in nums:
  for leter in 'a', 'b', 'c':
    for degree in '0.5654', '0.4564', '0.6546':
     for numbers in '6', '7', '8':
      print(num, leter, degree, numbers)
```

```python
#RANGE   (FOR LOPP)
for i in range(1, 11):
   print(i)
```

```python
#WHILE LOOP X += 1 IF THIS DIDN'T ADD WHOLE BECOME INFINITY
X = 0
while X < 10:
    print(X)
    X += 1
```

```python
#BREAK (WHILE LOOP)
x = 0

while x < 10:
    if x == 5:
        break

    print(x)
    x += 1
```
