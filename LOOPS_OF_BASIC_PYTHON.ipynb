```python
subjects = ('english', 'maths', 'science', 'hostory')
print(subjects)
```

```python
for item in subjects:
  print(item)
```

```python
for subject in enumerate(subjects):
  print(subjects)
```

```python
for index,subjects in enumerate(subjects):
  print(index,subjects)
```

```python
subjects_str = '_, _' . join(subjects)
print(subjects_str)
```

```python
subjects_str = '_,_' . join(subjects)
new_list = subjects_str.split('_,_')
print(subjects_str)
print(new_list)
```
