```python
#FOR CHARACTER/STRING IF & ELSE
language = 'python'
if language == 'python':
    print('conditional true')
else:
    print('no match')
```

```python
 #ELIF
 language = 'java'
if language == 'python':
    print('language is python')
elif language == 'java':
    print('language is java')
else:
    print('no match')
```

```python
#MULTIPLE ELIF
language = 'java'
if language == 'python':
    print('language is python')
elif language == 'java':
    print('language is java')
elif language == 'C++':
  print('language is c++')
elif language == 'javascript':
  print('language is javascript')
else:
    print('no match')
```

```python
#CONDITIONAL STATMENT (AND)
user = 'admin'
logged_in = False
if user == 'admin' and 'logged_in':
    print('ADMIN LOOGED IN WELCOME')
else:
    print('THIS IS CIA WE WILL CATCH U SOON!!')
```

```python
# (OR)
user = 'admin'
logged_in = True
if user == 'admin' or 'logged_in':
    print('ADMIN LOOGED IN WELCOME')
else:
    print('THIS IS CIA WE WILL CATCH U SOON!!!')
```

```python
#(NOT)
user = 'admin'
logged_in = False
if not logged_in:
    print('PLZ LOG IN ADMIN PAGE')
else:
    print('THIS IS CIA WE WILL CATCH U SOON!!!')
```

```python
#DIFFERENCE BETWEEN IS / ==
a = [1,2,3]
b = [1,2,3]
print(a == b)
```

```python
#DIFFERENCE BETWEEN IS / ==
a = [1,2,3]
b = [1,2,3]
print(a is b)
```

```python
#ID TRACKING OF BOTH LISTS
a = [1,2,3]
b = [1,2,3]
print(id(a))
print(id(b))
print(a == b)
print(a is b)
```

```python
#FOR SAME ID
a = [1,2,3]
b = a
print(id(a))
print(id(b))
print(id(a) == id(b))
```
